```python
from fpdf import FPDF

# Function to sanitize input text
def sanitize_text(text):
    # Replace problematic Unicode characters with ASCII equivalents
    replacements = {
        '’': "'", '“': '"', '”': '"', '–': '-', '—': '-', '…': '...'
    }
    for key, value in replacements.items():
        text = text.replace(key, value)
    return text

# Create the application letter in PDF format
pdf = FPDF()

pdf.set_auto_page_break(auto=True, margin=15)
pdf.add_page()
pdf.set_font("Arial", size=12)

# Title and Contact Information
pdf.set_font("Arial", style="B", size=14)
pdf.cell(0, 10, "ROSELINE IKOME", ln=True, align='C')
pdf.set_font("Arial", size=12)
pdf.cell(0, 10, "Block 2 Flat 2, 50 Unit Housing Estate", ln=True, align='C')
pdf.cell(0, 10, "Idu Uruan, Akwa Ibom State, Nigeria", ln=True, align='C')
pdf.cell(0, 10, "Phone: +234 816 367 4964 | Email: lienguikome@gmail.com", ln=True, align='C')
pdf.ln(20)

# Date
pdf.cell(0, 10, "[Date: Insert Current Date Here]", ln=True)
pdf.ln(10)

# Recipient Address
pdf.cell(0, 10, "Hiring Manager", ln=True)
pdf.cell(0, 10, "Federal Ministry of Agriculture and Food Security", ln=True)
pdf.cell(0, 10, "[Address: Add Here if Known]", ln=True)
pdf.ln(10)

# Subject Line
pdf.set_font("Arial", style="B", size=12)
pdf.cell(0, 10, "Dear Hiring Manager,", ln=True)
pdf.ln(5)

# Body of the Letter
pdf.set_font("Arial", size=12)
application_body = (
    "Application for the Position of Knowledge Management and Communications Officer\n\n"
    "I am writing to express my keen interest in the Knowledge Management and Communications Officer role "
    "at the Federal Ministry of Agriculture and Food Security. With over a decade of experience in education, "
    "media relations, and stakeholder engagement, coupled with my strong background in communication, I am confident "
    "in my ability to contribute to your mission of driving agricultural and rural development through effective "
    "knowledge dissemination and advocacy.\n\n"
    "In my previous role as Press and Media Assistant with the Ifeko Ijaye LGA Acting Executive Chairman, I managed "
    "public relations and developed key advocacy messages that enhanced stakeholder engagement. My tenure as Acting "
    "Principal at Mount Ephraim High School further honed my organizational and leadership skills, where I facilitated "
    "training programs and streamlined communication between various stakeholders. These roles, along with my teaching "
    "experience, have equipped me with a deep understanding of content development, documentation, and audience-focused "
    "communication strategies.\n\n"
    "Key highlights of my qualifications include:\n"
    "- A Bachelor's degree in Mass Communication and a Post Graduate Diploma in Education, reflecting a solid academic foundation in communication and knowledge management.\n"
    "- Proven ability to design and deliver impactful educational and advocacy campaigns, as demonstrated during my time as Head of Department for Art at Treasure Child Comprehensive Secondary School.\n"
    "- Proficiency in digital tools such as Microsoft Office and Zoom, enabling effective reporting, collaboration, and content creation.\n\n"
    "I am passionate about leveraging my skills to create a culture of learning and knowledge-sharing within your organization. "
    "I am particularly drawn to this role’s emphasis on fostering innovation and engaging stakeholders in policy dialogues, which aligns "
    "with my professional values and expertise.\n\n"
    "I would welcome the opportunity to discuss how my experience and skills align with the objectives of your organization. Thank you for "
    "considering my application. I look forward to the possibility of contributing to the impactful work of the Federal Ministry of Agriculture "
    "and Food Security.\n\n"
    "Yours sincerely,\n\n"
    "Roseline Ikome"
)

# Sanitize the text before adding it to the PDF
application_body = sanitize_text(application_body)

pdf.multi_cell(0, 10, application_body)
pdf.ln(10)

# Save PDF
# Save PDF to the current directory
output_path_letter = "Application_Letter_Roseline_Ikome.pdf"
pdf.output(output_path_letter)

output_path_letter
output_path_letter
```
